```python
import pandas as pd


# ============================================
# 1. LOAD DATASET
# ============================================

df = pd.read_csv("my_data.csv")


# ============================================
# 2. INSPECT / VIEW DATA
# ============================================

print("----- First 5 Rows -----")
print(df.head())

print("\n----- Dataset Information -----")
print(df.info())

print("\n----- Statistical Summary -----")
print(df.describe())


# ============================================
# 3. CHECK MISSING VALUES
# ============================================

print("\n----- Missing Values Before Cleaning -----")
print(df.isnull().sum())


# ============================================
# 4. REMOVE DUPLICATE ROWS
# ============================================

df = df.drop_duplicates()

print("\n----- Duplicate Rows Removed -----")
print("Remaining rows:", len(df))


# ============================================
# 5. FILL NUMERICAL MISSING VALUES
#    Using Median
# ============================================

df["age"] = df["age"].fillna(df["age"].median())

print("\n----- Missing Values in age Filled -----")


# ============================================
# 6. FILL CATEGORICAL MISSING VALUES
#    Using Mode
# ============================================

df["city"] = df["city"].fillna(df["city"].mode()[0])

print("----- Missing Values in city Filled -----")


# ============================================
# 7. CLEAN TEXT DATA
#    Remove Spaces + Convert to Lowercase
# ============================================

df["city"] = df["city"].str.strip().str.lower()

print("\n----- city Column Cleaned -----")


# ============================================
# 8. CONVERT age INTO NUMERICAL DATA TYPE
# ============================================

df["age"] = pd.to_numeric(df["age"], errors="coerce")

print("----- age Data Type Converted -----")


# ============================================
# 9. REMOVE INVALID age VALUES
#    Valid age: 15 to 60
# ============================================

df = df[(df["age"] >= 15) & (df["age"] <= 60)]

print("----- Invalid age Values Removed -----")


# ============================================
# 10. CHECK MISSING VALUES AFTER CLEANING
# ============================================

print("\n----- Missing Values After Cleaning -----")
print(df.isnull().sum())


# ============================================
# 11. CHECK DUPLICATE VALUES AFTER CLEANING
# ============================================

print("\n----- Duplicate Rows After Cleaning -----")
print(df.duplicated().sum())


# ============================================
# 12. VIEW CLEANED DATA
# ============================================

print("\n----- Cleaned Dataset -----")
print(df.head())


# ============================================
# 13. SAVE CLEANED DATASET
# ============================================

df.to_csv("cleaned_student.csv", index=False)

print("\n----- Data Cleaning Completed Successfully! -----")
```
