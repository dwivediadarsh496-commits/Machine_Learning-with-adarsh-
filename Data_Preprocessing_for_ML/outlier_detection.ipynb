# 🔵 PART 1 — IQR METHOD

## 🟢 Cell 13 — IQR Concept

## IQR Method

IQR stands for Interquartile Range.

**IQR = Q3 - Q1**

**Lower Bound = Q1 - 1.5 × IQR**

**Upper Bound = Q3 + 1.5 × IQR**

Any value below the lower bound or above the upper bound is considered a potential outlier.

## 🟢 Cell 14 — Calculate IQR for Age

````python
# ============================================
# IQR METHOD FOR AGE
# ============================================

Q1_age = df["age"].quantile(0.25)
Q3_age = df["age"].quantile(0.75)

IQR_age = Q3_age - Q1_age

lower_age = Q1_age - 1.5 * IQR_age
upper_age = Q3_age + 1.5 * IQR_age

print("Q1:", Q1_age)
print("Q3:", Q3_age)
print("IQR:", IQR_age)
print("Lower Bound:", lower_age)
print("Upper Bound:", upper_age)
````

## 🟢 Cell 15 — Detect Age Outliers

````python
# ============================================
# DETECT AGE OUTLIERS
# ============================================

age_outliers = df[
    (df["age"] < lower_age) |
    (df["age"] > upper_age)
]

print("Age Outliers:")
print(age_outliers)
````

**Important:**

- `(df["age"] < lower_age)` means lower bound se chhota.
- `(df["age"] > upper_age)` means upper bound se bada.
- `|` means OR.

## 🟢 Cell 16 — Count Age Outliers

````python
# ============================================
# COUNT AGE OUTLIERS
# ============================================

print("Number of Age Outliers:", len(age_outliers))
````

## 🟢 Cell 17 — IQR for Marks

````python
# ============================================
# IQR METHOD FOR MARKS
# ============================================

Q1_marks = df["marks"].quantile(0.25)
Q3_marks = df["marks"].quantile(0.75)

IQR_marks = Q3_marks - Q1_marks

lower_marks = Q1_marks - 1.5 * IQR_marks
upper_marks = Q3_marks + 1.5 * IQR_marks

print("Q1:", Q1_marks)
print("Q3:", Q3_marks)
print("IQR:", IQR_marks)
print("Lower Bound:", lower_marks)
print("Upper Bound:", upper_marks)
````

## 🟢 Cell 18 — Detect Marks Outliers

````python
# ============================================
# DETECT MARKS OUTLIERS
# ============================================

marks_outliers = df[
    (df["marks"] < lower_marks) |
    (df["marks"] > upper_marks)
]

print("Marks Outliers:")
print(marks_outliers)
````

## 🟢 Cell 19 — Count Marks Outliers

````python
# ============================================
# COUNT MARKS OUTLIERS
# ============================================

print("Number of Marks Outliers:", len(marks_outliers))
````

## 🟢 Cell 20 — Generic IQR Function

Ab ek reusable function banate hain.

````python
# ============================================
# GENERIC IQR OUTLIER FUNCTION
# ============================================

def detect_outliers_iqr(df, column):

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    return outliers, lower_bound, upper_bound
````

Ab kisi bhi numerical column par use kar sakte ho:

````python
# Age
age_outliers, lower, upper = detect_outliers_iqr(df, "age")

print(age_outliers)
````

````python
# Marks
marks_outliers, lower, upper = detect_outliers_iqr(df, "marks")

print(marks_outliers)
````

# 🔵 PART 2 — Z-SCORE METHOD

## 🟢 Cell 21 — Z-Score Concept

## Z-Score Method

Z-score tells us how many standard deviations a data point is away from the mean.

Formula:

**Z = (X - Mean) / Standard Deviation**

Generally:

**|Z| > 3**

is considered a potential outlier.

## 🟢 Cell 22 — Calculate Z-Score

````python
# ============================================
# CALCULATE Z-SCORE FOR AGE
# ============================================

df["age_zscore"] = zscore(df["age"])

df[["age", "age_zscore"]]
````

## 🟢 Cell 23 — Detect Age Outliers Using Z-Score

````python
# ============================================
# DETECT AGE OUTLIERS USING Z-SCORE
# ============================================

age_zscore_outliers = df[
    abs(df["age_zscore"]) > 3
]

print("Age Outliers using Z-Score:")
print(age_zscore_outliers)
````

**`abs()` kya karta hai?**

Suppose `-3.5` aur `+3.5` — dono ko `3.5` consider karega.

Isliye `abs(df["age_zscore"]) > 3` likhte hain.

## 🟢 Cell 24 — Z-Score for Marks

````python
# ============================================
# CALCULATE Z-SCORE FOR MARKS
# ============================================

df["marks_zscore"] = zscore(df["marks"])

df[["marks", "marks_zscore"]]
````

## 🟢 Cell 25 — Detect Marks Outliers

````python
# ============================================
# DETECT MARKS OUTLIERS USING Z-SCORE
# ============================================

marks_zscore_outliers = df[
    abs(df["marks_zscore"]) > 3
]

print("Marks Outliers using Z-Score:")
print(marks_zscore_outliers)
````

# 🔵 PART 3 — REMOVE OUTLIERS

⚠️ **Important:** Real-world dataset mein automatically outlier delete nahi karna chahiye. Pehle check karo ki value genuine hai ya data-entry error.

## 🟢 Cell 26 — Remove Age Outliers Using IQR

````python
# ============================================
# REMOVE AGE OUTLIERS
# ============================================

df_cleaned = df[
    (df["age"] >= lower_age) &
    (df["age"] <= upper_age)
].copy()

print("Original Rows:", len(df))
print("Rows After Removing Age Outliers:", len(df_cleaned))
````

## 🟢 Cell 27 — Remove Marks Outliers

Agar age aur marks dono ke outliers remove karne hain:

````python
# ============================================
# REMOVE AGE AND MARKS OUTLIERS
# ============================================

df_cleaned = df[
    (df["age"] >= lower_age) &
    (df["age"] <= upper_age) &
    (df["marks"] >= lower_marks) &
    (df["marks"] <= upper_marks)
].copy()

print("Original Rows:", len(df))
print("Cleaned Rows:", len(df_cleaned))
````

## ⭐ Bonus — Automatically Detect Outliers in All Numerical Columns

Ye real projects mein kaafi useful hai.

````python
# ============================================
# DETECT OUTLIERS IN ALL NUMERICAL COLUMNS
# ============================================

numerical_columns = df.select_dtypes(include="number").columns

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print("\nColumn:", column)
    print("Number of Outliers:", len(outliers))
````

**Note:** Isme `student_id` bhi numerical hai, lekin ID ko generally outlier analysis mein include nahi karte. Better:

````python
numerical_columns = ["age", "marks"]

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print("\nColumn:", column)
    print("Outliers:", len(outliers))
````
