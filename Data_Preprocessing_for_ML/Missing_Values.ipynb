```python
import pandas as pd

# ============================================
# 1. LOAD DATASET
# ============================================

df = pd.read_csv("my_data.csv")


# ============================================
# 2. CHECK MISSING VALUES
# ============================================

print("----- Missing Values Before Cleaning -----")
print(df.isnull().sum())


# ============================================
# 3. CHECK MISSING VALUE PERCENTAGE
# ============================================

print("\n----- Missing Value Percentage -----")
print(df.isnull().mean() * 100)


# ============================================
# 4. FILL NUMERICAL MISSING VALUES
#    Using Median
# ============================================

df["age"] = df["age"].fillna(df["age"].median())


# ============================================
# 5. FILL CATEGORICAL MISSING VALUES
#    Using Mode
# ============================================

df["city"] = df["city"].fillna(df["city"].mode()[0])


# ============================================
# 6. CHECK MISSING VALUES AFTER CLEANING
# ============================================

print("\n----- Missing Values After Cleaning -----")
print(df.isnull().sum())


# ============================================
# 7. SAVE CLEANED DATASET
# ============================================

df.to_csv("student_cleaned.csv", index=False)

print("\nMissing Value Handling Completed!")
```
