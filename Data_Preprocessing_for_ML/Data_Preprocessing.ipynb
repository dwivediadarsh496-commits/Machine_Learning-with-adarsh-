# Data collection

Data Collection is the process of gathering relevant data from various sources for analysis, research, or machine learning

```python
import pandas as pd 

data = pd.read_csv("my_data.csv")

print(data.head())

print("\nshape of data \n")
print("shape:", data.shape)

print("colums::\n")
print("colums:" , data.columns)

print("data type::\n")
print(data.dtypes)

print("\nmissing value")
print(data.isnull().sum())

print(data.info())

print("discribe:")
print(data.describe())
```
