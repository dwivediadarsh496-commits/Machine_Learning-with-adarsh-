# 📓 Label Encoding — Complete Jupyter Notebook

Is notebook mein hum `LabelEncoder` se categorical (text) data ko numbers mein convert karna seekhenge.

## 🟢 Cell 1 — Import Libraries

````python
# ============================================
# IMPORT LIBRARIES
# ============================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder

print("Libraries imported successfully!")
````

## 🟢 Cell 2 — Create Dataset

````python
# ============================================
# CREATE SAMPLE DATASET
# ============================================

data = {
    "student": ["A", "B", "C", "D", "E", "F"],
    "city": ["Delhi", "Mumbai", "Bhilai", "Delhi", "Mumbai", "Bhilai"],
    "marks": [80, 75, 90, 85, 70, 88]
}

df = pd.DataFrame(data)

df
````

Expected output:

```
  student    city  marks
0       A   Delhi     80
1       B  Mumbai     75
2       C  Bhilai     90
3       D   Delhi     85
4       E  Mumbai     70
5       F  Bhilai     88
```

## 🟢 Cell 3 — Check Data Types

````python
# ============================================
# CHECK DATA TYPES
# ============================================

print(df.dtypes)
````

`city` ka type generally:

```
object
```

hoga. Matlab categorical/text data.

## 🟢 Cell 4 — Create Label Encoder

````python
# ============================================
# CREATE LABEL ENCODER
# ============================================

encoder = LabelEncoder()

encoder
````

Yahan `encoder` ek `LabelEncoder` object hai jo categories ko numbers mein convert karega.

## 🟢 Cell 5 — Apply Label Encoding

````python
# ============================================
# APPLY LABEL ENCODING TO CITY
# ============================================

df["city_encoded"] = encoder.fit_transform(df["city"])

df
````

Possible output:

```
  student    city  marks  city_encoded
0       A   Delhi     80             1
1       B  Mumbai     75             2
2       C  Bhilai     90             0
3       D   Delhi     85             1
4       E  Mumbai     70             2
5       F  Bhilai     88             0
```

### 🔥 `fit_transform()` kya karta hai?

Ye bahut important hai.

```python
encoder.fit_transform(df["city"])
```

Actually do kaam karta hai:

**`fit()`** — Categories ko learn karta hai:

```
Bhilai
Delhi
Mumbai
```

**`transform()`** — Unko numbers mein convert karta hai:

```
Bhilai → 0
Delhi  → 1
Mumbai → 2
```

So:

```
fit_transform()
     ↓
Learn + Convert
```

## 🟢 Cell 6 — Check Mapping

````python
# ============================================
# CHECK CATEGORY MAPPING
# ============================================

print("Classes:")
print(encoder.classes_)
````

Output:

```
['Bhilai' 'Delhi' 'Mumbai']
```

Mapping:

```
Bhilai → 0
Delhi  → 1
Mumbai → 2
```

## 🟢 Cell 7 — Manually See Mapping

````python
# ============================================
# DISPLAY CATEGORY AND LABEL
# ============================================

for category, label in zip(encoder.classes_, encoder.transform(encoder.classes_)):
    print(category, "→", label)
````

Output:

```
Bhilai → 0
Delhi → 1
Mumbai → 2
```

## 🟢 Cell 8 — Convert Numbers Back

`LabelEncoder` inverse transformation bhi kar sakta hai.

````python
# ============================================
# CONVERT LABELS BACK TO ORIGINAL CATEGORIES
# ============================================

df["city_original"] = encoder.inverse_transform(df["city_encoded"])

df
````

Example:

```
0 → Bhilai
1 → Delhi
2 → Mumbai
```

Isko kehte hain `inverse_transform()`.

## 📌 Complete Basic Code

Agar exam/practical mein sirf Label Encoding implement karna ho:

````python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Create dataset
df = pd.DataFrame({
    "city": ["Delhi", "Mumbai", "Bhilai", "Delhi", "Mumbai"]
})

# Create encoder
encoder = LabelEncoder()

# Apply label encoding
df["city_encoded"] = encoder.fit_transform(df["city"])

# Display result
print(df)

# Display mapping
print("\nMapping:")
for category, label in zip(
    encoder.classes_,
    encoder.transform(encoder.classes_)
):
    print(category, "→", label)
````
